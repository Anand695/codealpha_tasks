```python
# ================================================================
# STUDENT PERFORMANCE FACTORS
# COMPLETE EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================
#
# This project is designed according to the EDA requirements:
#
# 1. Ask meaningful questions before analysis
# 2. Explore data structure and data types
# 3. Identify trends, patterns and anomalies
# 4. Test hypotheses using statistics
# 5. Validate observations using visualization
# 6. Detect data issues/problems
# 7. Compare Top 10% and Bottom 10% students
#
# ================================================================


# ================================================================
# 1. PROJECT TITLE
# ================================================================

from IPython.display import display, Markdown

display(Markdown("""
# 📊 Student Academic Performance Analysis

## Exploratory Data Analysis using Python

### Objective

To explore how academic habits, attendance, previous performance,
sleep, tutoring, physical activity and other student-related factors
are associated with examination performance.
"""))


# ================================================================
# 2. MEANINGFUL QUESTIONS BEFORE ANALYSIS
# ================================================================

display(Markdown("""
## 2. Meaningful Questions

Before analyzing the dataset, the following questions are defined:

### Academic Questions
1. Does studying for more hours relate to a higher exam score?
2. Does higher attendance relate to better examination performance?
3. Is previous academic performance strongly related to the current exam score?
4. Do tutoring sessions show an association with exam performance?

### Lifestyle Questions
5. Is sleep duration associated with exam performance?
6. Is physical activity associated with exam performance?

### Comparative Questions
7. How different are the habits of the Top 10% and Bottom 10% students?
8. Which factors show the strongest correlation with Exam Score?

### Data Quality Questions
9. Are there missing values?
10. Are there duplicate records or unusual/outlier observations?
"""))

print("""
Observation:
These questions define the direction of the exploratory analysis
before examining the results, helping prevent random or irrelevant
analysis.
""")


# ================================================================
# 3. IMPORT LIBRARIES
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, ttest_ind

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")

print("""
Observation:
Pandas is used for data handling, NumPy for numerical operations,
Matplotlib and Seaborn for visualization, and SciPy for statistical
hypothesis testing.
""")


# ================================================================
# 4. LOAD DATASET
# ================================================================

# ------------------------------------------------
# CHANGE ONLY THIS LINE IF YOUR FILE HAS A DIFFERENT NAME
# ------------------------------------------------

FILE_NAME = "StudentPerformanceFactors.csv"

df = pd.read_csv(FILE_NAME)

print("Dataset loaded successfully.")
print()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nFirst 5 records:")
display(df.head())

print("""
Observation:
The dataset has been successfully loaded. The first few records
provide an initial view of the information collected for each
student.
""")


# ================================================================
# 5. DATASET STRUCTURE AND COLUMN INFORMATION
# ================================================================

display(Markdown("""
## 5. Dataset Structure and Variables
"""))

print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
display(
    df.dtypes.to_frame("Data Type")
)

print("\nDataset Information:")
df.info()

print("""
Observation:
The dataset contains both numerical and categorical variables.
Numerical variables can be used for statistical analysis and
correlation, while categorical variables can be explored through
group comparisons.
""")


# ================================================================
# 6. DATA QUALITY CHECK
# ================================================================

display(Markdown("""
## 6. Data Quality Check

The dataset is checked for:

- Missing values
- Duplicate records
- Invalid numerical values
- Overall data completeness
"""))


# Missing values

missing_values = df.isnull().sum()

print("Missing Values:")

if missing_values.sum() == 0:

    print("No missing values were found.")

else:

    display(
        missing_values[
            missing_values > 0
        ].to_frame("Missing Values")
    )


# Duplicate records

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)


# Numerical summary

print("\nBasic Statistical Summary:")

display(
    df.describe().round(2)
)


print("""
Observation:
This quality check helps determine whether the dataset contains
missing or duplicate records and provides an initial understanding
of the numerical variables.
""")


# ================================================================
# 7. DATA CLEANING
# ================================================================

display(Markdown("""
## 7. Data Cleaning
"""))


# Remove duplicate rows

before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

removed_duplicates = (
    before_duplicates - after_duplicates
)


print(
    "Duplicate rows removed:",
    removed_duplicates
)


# Handle numerical missing values

numeric_columns = df.select_dtypes(
    include=np.number
).columns


for column in numeric_columns:

    if df[column].isnull().sum() > 0:

        df[column] = df[column].fillna(
            df[column].median()
        )


# Handle categorical missing values

categorical_columns = df.select_dtypes(
    include="object"
).columns


for column in categorical_columns:

    if df[column].isnull().sum() > 0:

        df[column] = df[column].fillna(
            df[column].mode()[0]
        )


print("\nMissing values after cleaning:")

remaining_missing = df.isnull().sum()

if remaining_missing.sum() == 0:

    print("No missing values remain.")

else:

    display(
        remaining_missing[
            remaining_missing > 0
        ].to_frame("Remaining Missing Values")
    )


print("""
Observation:
The dataset has been cleaned by removing duplicate records and
handling missing values where necessary. The cleaned dataset is
now ready for exploratory analysis.
""")


# ================================================================
# 8. EXAM SCORE DISTRIBUTION
# ================================================================

display(Markdown("""
## 8. Distribution of Exam Scores
"""))


plt.figure(figsize=(9, 5))

sns.histplot(
    df["Exam_Score"],
    bins=20,
    kde=True
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


mean_score = df["Exam_Score"].mean()
median_score = df["Exam_Score"].median()
std_score = df["Exam_Score"].std()

print(f"Mean Exam Score: {mean_score:.2f}")
print(f"Median Exam Score: {median_score:.2f}")
print(f"Standard Deviation: {std_score:.2f}")

print(f"""
Observation:
The average examination score is {mean_score:.2f}, while the
median is {median_score:.2f}. The distribution shows how student
performance varies across the dataset.
""")


# ================================================================
# 9. STUDY HOURS DISTRIBUTION
# ================================================================

display(Markdown("""
## 9. Distribution of Hours Studied
"""))


plt.figure(figsize=(9, 5))

sns.histplot(
    df["Hours_Studied"],
    bins=15,
    kde=True
)

plt.title("Distribution of Hours Studied")
plt.xlabel("Hours Studied")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


print(
    f"Average Hours Studied: "
    f"{df['Hours_Studied'].mean():.2f}"
)

print("""
Observation:
Students show variation in the amount of time spent studying.
This variable can now be compared with examination performance.
""")


# ================================================================
# 10. STUDY HOURS VS EXAM SCORE
# ================================================================

display(Markdown("""
## 10. Study Hours vs Exam Score
"""))


study_corr, study_p = pearsonr(
    df["Hours_Studied"],
    df["Exam_Score"]
)


plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x="Hours_Studied",
    y="Exam_Score",
    scatter_kws={"alpha": 0.35}
)

plt.title("Hours Studied vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()


print(
    f"Pearson Correlation: {study_corr:.3f}"
)

print(
    f"P-value: {study_p:.6f}"
)

print(f"""
Observation:
Hours Studied has a Pearson correlation of {study_corr:.3f}
with Exam Score. The scatter plot and regression line show the
direction of the relationship.

The p-value is {study_p:.6f}, which can be used to determine whether
the observed correlation is statistically significant.
""")


# ================================================================
# 11. ATTENDANCE VS EXAM SCORE
# ================================================================

display(Markdown("""
## 11. Attendance vs Exam Score
"""))


attendance_corr, attendance_p = pearsonr(
    df["Attendance"],
    df["Exam_Score"]
)


plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x="Attendance",
    y="Exam_Score",
    scatter_kws={"alpha": 0.35}
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()


print(
    f"Pearson Correlation: {attendance_corr:.3f}"
)

print(
    f"P-value: {attendance_p:.6f}"
)

print(f"""
Observation:
Attendance has a Pearson correlation of {attendance_corr:.3f}
with Exam Score. The visualization shows whether higher attendance
is associated with better examination performance.
""")


# ================================================================
# 12. PREVIOUS SCORES VS EXAM SCORE
# ================================================================

display(Markdown("""
## 12. Previous Scores vs Exam Score
"""))


previous_corr, previous_p = pearsonr(
    df["Previous_Scores"],
    df["Exam_Score"]
)


plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x="Previous_Scores",
    y="Exam_Score",
    scatter_kws={"alpha": 0.35}
)

plt.title("Previous Scores vs Exam Score")
plt.xlabel("Previous Scores")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()


print(
    f"Pearson Correlation: {previous_corr:.3f}"
)

print(
    f"P-value: {previous_p:.6f}"
)

print(f"""
Observation:
Previous Scores have a Pearson correlation of {previous_corr:.3f}
with Exam Score. This relationship helps determine whether earlier
academic performance is associated with current performance.
""")


# ================================================================
# 13. SLEEP HOURS VS EXAM SCORE
# ================================================================

display(Markdown("""
## 13. Sleep Hours vs Exam Score
"""))


sleep_corr, sleep_p = pearsonr(
    df["Sleep_Hours"],
    df["Exam_Score"]
)


plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x="Sleep_Hours",
    y="Exam_Score",
    scatter_kws={"alpha": 0.35}
)

plt.title("Sleep Hours vs Exam Score")
plt.xlabel("Sleep Hours")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()


print(
    f"Pearson Correlation: {sleep_corr:.3f}"
)

print(
    f"P-value: {sleep_p:.6f}"
)

print(f"""
Observation:
Sleep Hours have a Pearson correlation of {sleep_corr:.3f}
with Exam Score. The graph helps examine whether sleep duration
has an observable relationship with academic performance.
""")


# ================================================================
# 14. TUTORING SESSIONS VS EXAM SCORE
# ================================================================

display(Markdown("""
## 14. Tutoring Sessions vs Exam Score
"""))


tutoring_corr, tutoring_p = pearsonr(
    df["Tutoring_Sessions"],
    df["Exam_Score"]
)


plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x="Tutoring_Sessions",
    y="Exam_Score",
    scatter_kws={"alpha": 0.35}
)

plt.title("Tutoring Sessions vs Exam Score")
plt.xlabel("Tutoring Sessions")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()


print(
    f"Pearson Correlation: {tutoring_corr:.3f}"
)

print(
    f"P-value: {tutoring_p:.6f}"
)

print(f"""
Observation:
Tutoring Sessions have a Pearson correlation of
{tutoring_corr:.3f} with Exam Score. The graph shows whether
additional tutoring is associated with examination performance.
""")


# ================================================================
# 15. PHYSICAL ACTIVITY VS EXAM SCORE
# ================================================================

display(Markdown("""
## 15. Physical Activity vs Exam Score
"""))


physical_corr, physical_p = pearsonr(
    df["Physical_Activity"],
    df["Exam_Score"]
)


plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x="Physical_Activity",
    y="Exam_Score",
    scatter_kws={"alpha": 0.35}
)

plt.title("Physical Activity vs Exam Score")
plt.xlabel("Physical Activity")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()


print(
    f"Pearson Correlation: {physical_corr:.3f}"
)

print(
    f"P-value: {physical_p:.6f}"
)

print(f"""
Observation:
Physical Activity has a Pearson correlation of
{physical_corr:.3f} with Exam Score. The analysis examines
whether physical activity shows an association with performance.
""")


# ================================================================
# 16. CORRELATION HEATMAP
# ================================================================

display(Markdown("""
## 16. Correlation Heatmap

The heatmap provides an overall view of relationships between
numerical variables in the dataset.
"""))


numeric_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()


plt.figure(figsize=(13, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")

plt.tight_layout()
plt.show()


print("""
Observation:
The heatmap provides a visual summary of positive and negative
relationships between numerical variables. Values closer to +1
or -1 represent stronger linear relationships, while values near
0 represent weaker linear relationships.
""")


# ================================================================
# 17. CORRELATION RANKING WITH EXAM SCORE
# ================================================================

display(Markdown("""
## 17. Factors Ranked by Correlation with Exam Score
"""))


exam_correlations = (
    correlation_matrix["Exam_Score"]
    .drop("Exam_Score")
    .sort_values(
        key=abs,
        ascending=False
    )
)


correlation_table = pd.DataFrame({
    "Correlation with Exam Score":
        exam_correlations
})


display(
    correlation_table.round(3)
)


strongest_factor = exam_correlations.index[0]
strongest_value = exam_correlations.iloc[0]


print(f"""
Observation:
Among the numerical variables analyzed, {strongest_factor} shows
the strongest linear correlation with Exam Score, with a
correlation of {strongest_value:.3f}.

This identifies the variables that deserve greater attention in
further analysis.
""")


# ================================================================
# 18. OUTLIER / ANOMALY DETECTION
# ================================================================

display(Markdown("""
## 18. Outlier and Anomaly Detection

The Interquartile Range (IQR) method is used to identify unusually
low or high observations.
"""))


outlier_columns = [
    "Hours_Studied",
    "Attendance",
    "Sleep_Hours",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Physical_Activity",
    "Exam_Score"
]


outlier_results = []


for column in outlier_columns:

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({

        "Variable": column,

        "Outliers": len(outliers),

        "Percentage":
            (len(outliers) / len(df)) * 100

    })


outlier_table = pd.DataFrame(
    outlier_results
)


outlier_table["Percentage"] = (
    outlier_table["Percentage"]
    .round(2)
)


display(outlier_table)


print("""
Observation:
The outlier table identifies observations that fall unusually far
from the middle 50% of the data. These observations should be
examined rather than automatically removed because an outlier may
represent either a genuine student characteristic or a data issue.
""")


# ================================================================
# 19. TOP 10% VS BOTTOM 10%
# ================================================================

display(Markdown("""
## 19. Top 10% vs Bottom 10% Students

Students are divided according to Exam Score:

- Top 10% = students at or above the 90th percentile
- Bottom 10% = students at or below the 10th percentile

Their academic habits and characteristics are then compared.
"""))


top_threshold = df["Exam_Score"].quantile(0.90)

bottom_threshold = df["Exam_Score"].quantile(0.10)


top_students = df[
    df["Exam_Score"] >= top_threshold
]


bottom_students = df[
    df["Exam_Score"] <= bottom_threshold
]


print(
    f"Top 10% threshold: {top_threshold:.2f}"
)

print(
    f"Bottom 10% threshold: {bottom_threshold:.2f}"
)

print(
    f"Number of Top 10% students: "
    f"{len(top_students)}"
)

print(
    f"Number of Bottom 10% students: "
    f"{len(bottom_students)}"
)


comparison_columns = [
    "Hours_Studied",
    "Attendance",
    "Sleep_Hours",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Physical_Activity",
    "Exam_Score"
]


top_bottom_comparison = pd.DataFrame({

    "Top 10%":
        top_students[
            comparison_columns
        ].mean(),

    "Bottom 10%":
        bottom_students[
            comparison_columns
        ].mean()

})


top_bottom_comparison["Difference"] = (
    top_bottom_comparison["Top 10%"]
    -
    top_bottom_comparison["Bottom 10%"]
)


display(
    top_bottom_comparison.round(2)
)


print("""
Observation:
The table compares the average academic and lifestyle
characteristics of the highest-performing and lowest-performing
students. Larger differences indicate factors where the two groups
show greater separation.
""")


# ================================================================
# 20. TOP 10% VS BOTTOM 10% VISUALIZATION
# ================================================================

display(Markdown("""
## 20. Top 10% vs Bottom 10% Visualization
"""))


plot_data = top_bottom_comparison.drop(
    index="Exam_Score"
)


plot_data[
    ["Top 10%", "Bottom 10%"]
].plot(
    kind="bar",
    figsize=(12, 6)
)


plt.title(
    "Top 10% vs Bottom 10% Student Characteristics"
)

plt.xlabel("Factors")

plt.ylabel("Average Value")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.legend(
    title="Student Group"
)

plt.tight_layout()

plt.show()


print("""
Observation:
The bar chart provides a visual comparison between the habits and
characteristics of the Top 10% and Bottom 10% students.
""")


# ================================================================
# 21. HYPOTHESIS TESTING
# ================================================================

display(Markdown("""
## 21. Hypothesis Testing

For each selected numerical factor:

### Null Hypothesis (H₀)
There is no statistically significant linear relationship between
the selected factor and Exam Score.

### Alternative Hypothesis (H₁)
There is a statistically significant linear relationship between
the selected factor and Exam Score.

A significance level of 0.05 is used.
"""))


hypothesis_variables = {

    "Hours Studied":
        "Hours_Studied",

    "Attendance":
        "Attendance",

    "Previous Scores":
        "Previous_Scores",

    "Sleep Hours":
        "Sleep_Hours",

    "Tutoring Sessions":
        "Tutoring_Sessions",

    "Physical Activity":
        "Physical_Activity"
}


hypothesis_results = []


for display_name, column in hypothesis_variables.items():

    r, p = pearsonr(
        df[column],
        df["Exam_Score"]
    )

    if p < 0.05:

        result = "Reject H₀"

        interpretation = (
            "Statistically significant"
        )

    else:

        result = "Fail to Reject H₀"

        interpretation = (
            "Not statistically significant"
        )


    hypothesis_results.append({

        "Factor":
            display_name,

        "Correlation":
            r,

        "P-value":
            p,

        "Decision":
            result,

        "Interpretation":
            interpretation

    })


hypothesis_table = pd.DataFrame(
    hypothesis_results
)


display(
    hypothesis_table.round(4)
)


print("""
Observation:
A p-value below 0.05 indicates statistically significant evidence
of a linear association with Exam Score in this dataset.

A statistically significant relationship does NOT prove causation.
It only indicates that the observed association is unlikely to be
explained by random sampling variation alone under the null hypothesis.
""")


# ================================================================
# 22. STRONGEST POSITIVE AND NEGATIVE RELATIONSHIPS
# ================================================================

display(Markdown("""
## 22. Strongest Positive and Negative Relationships
"""))


positive_correlations = exam_correlations[
    exam_correlations > 0
]


negative_correlations = exam_correlations[
    exam_correlations < 0
]


print("Positive relationships with Exam Score:")

display(
    positive_correlations
    .sort_values(
        ascending=False
    )
    .to_frame(
        "Correlation"
    )
    .round(3)
)


print("\nNegative relationships with Exam Score:")

if len(negative_correlations) > 0:

    display(
        negative_correlations
        .sort_values()
        .to_frame(
            "Correlation"
        )
        .round(3)
    )

else:

    print(
        "No negative correlations were found."
    )


print("""
Observation:
Positive correlations indicate that the two variables tend to
increase together, while negative correlations indicate an
opposite direction of association.
""")


# ================================================================
# 23. TOP 10% VS BOTTOM 10% DIFFERENCE RANKING
# ================================================================

display(Markdown("""
## 23. Factors with the Largest Top 10% vs Bottom 10% Difference
"""))


difference_ranking = (
    top_bottom_comparison["Difference"]
    .drop("Exam_Score")
    .abs()
    .sort_values(
        ascending=False
    )
)


difference_table = pd.DataFrame({

    "Absolute Difference":
        difference_ranking

})


display(
    difference_table.round(2)
)


largest_difference_factor = (
    difference_ranking.index[0]
)


largest_difference = (
    difference_ranking.iloc[0]
)


print(f"""
Observation:
{largest_difference_factor} shows the largest average difference
between the Top 10% and Bottom 10% groups, with an absolute
difference of approximately {largest_difference:.2f}.
""")


# ================================================================
# 24. OVERALL FINDINGS
# ================================================================

display(Markdown("""
## 24. Overall Findings
"""))


print("=" * 70)
print("                    KEY FINDINGS")
print("=" * 70)


print(
    f"""
1. Strongest numerical correlation with Exam Score:
   {strongest_factor}
   Correlation = {strongest_value:.3f}

2. Hours Studied correlation:
   {study_corr:.3f}

3. Attendance correlation:
   {attendance_corr:.3f}

4. Previous Scores correlation:
   {previous_corr:.3f}

5. Sleep Hours correlation:
   {sleep_corr:.3f}

6. Tutoring Sessions correlation:
   {tutoring_corr:.3f}

7. Physical Activity correlation:
   {physical_corr:.3f}

8. Largest Top 10% vs Bottom 10% difference:
   {largest_difference_factor}
   Difference = {largest_difference:.2f}
"""
)


# ================================================================
# 25. FINAL CONCLUSION
# ================================================================

display(Markdown("""
## 25. Final Conclusion

### Conclusion

This exploratory data analysis investigated the factors associated
with student examination performance.

The analysis explored:

- Dataset structure and data types
- Missing values and duplicate records
- Statistical summaries
- Score distributions
- Academic habits
- Study hours
- Attendance
- Previous academic performance
- Sleep duration
- Tutoring sessions
- Physical activity
- Correlation relationships
- Outliers and anomalies
- Hypothesis testing
- Top 10% vs Bottom 10% student characteristics

The correlation analysis identifies which numerical variables have
stronger or weaker linear associations with Exam Score.

The Top 10% vs Bottom 10% comparison provides an additional way
to identify differences in the characteristics of high-performing
and low-performing students.

### Important Limitation

The analysis identifies associations rather than proving cause and
effect. A strong correlation does not necessarily mean that one
factor directly causes a change in examination performance.

The findings should therefore be interpreted as patterns observed
within this dataset.
"""))


# ================================================================
# 26. PROJECT COMPLETION
# ================================================================

print("\n")
print("=" * 70)
print("       STUDENT PERFORMANCE FACTORS - EDA")
print("                 PROJECT COMPLETED")
print("=" * 70)

print("""
The project has completed the major requirements of Exploratory
Data Analysis:

✓ Meaningful questions
✓ Data structure exploration
✓ Data cleaning
✓ Descriptive statistics
✓ Trends and patterns
✓ Correlation analysis
✓ Visualization
✓ Anomaly detection
✓ Hypothesis testing
✓ Top 10% vs Bottom 10% comparison
✓ Final findings and conclusion
""")
```
